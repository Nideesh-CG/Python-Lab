```python
import threading
import time

start=time.perf_counter()

def do_something(seconds): # simple functions with argument
    print(f'sleeping {seconds} second(s)')
    time.sleep(seconds)
    print("Done something...")

threads=[]

for _ in range(10):
    t=threading.Thread(target=do_something,args=[1.2])
    t.start()
    threads.append(t)
    
for thread in threads:
    thread.join()

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} seconds')# printing the thread execution time
```

```python
import concurrent.futures
import time

start=time.perf_counter()

def do_something(seconds): # simple functions with argument
    print(f'sleeping {seconds} second(s)')
    time.sleep(seconds) #sleep function of thread 
    return 'Done something...'

#using concurent.futures module 
#Using ThreadPoolExecutor 

with concurrent.futures.ThreadPoolExecutor() as exe:
    f1=exe.submit(do_something,1.5) # submit() function stores the future object
    print(f1.result())

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} seconds')# printing the thread execution time
```

```python
import concurrent.futures
import time

start=time.perf_counter()

def do_something(seconds): # simple functions with argument
    print(f'sleeping {seconds} second(s)')
    time.sleep(seconds) #sleep function of thread 
    return 'Done something...'

#using concurent.futures module 
#Using ThreadPoolExecutor 

with concurrent.futures.ThreadPoolExecutor() as exe:
    
    res=[exe.submit(do_something,1.5) for _ in range(2)]# creating mutiple threads using list comprehension
    for f in concurrent.futures.as_completed(res):  # concurrent.futures.as_completed() takes  list of future object and returns iterators 
         print(f.result())

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} seconds') # printing the thread execution time
```

```python
# take series of arguments 
import concurrent.futures
import time

start=time.perf_counter()

def do_something(seconds): # simple functions with argument
    print(f'sleeping {seconds} second(s)')
    time.sleep(seconds) #sleep function of thread 
    return 'Done something...'

#using concurent.futures module 
#Using ThreadPoolExecutor 

with concurrent.futures.ThreadPoolExecutor() as exe:
    secs=[5,4,3,2,1] # series of arguments 
    res=[exe.submit(do_something,sec) for sec in secs]# creating mutiple threads using list comprehension
    for f in concurrent.futures.as_completed(res):  # concurrent.futures.as_completed() takes  list of future object and returns iterators 
         print(f.result())

finish=time.perf_counter()

print(f'Finished in {round(finish-start,2)} seconds') # printing the thread execution time
```

```python
import time
import requests
import concurrent.futures

audio_files=['https://commondatastorage.googleapis.com/codeskulptor-assets/sounddogs/missile.mp3',
            'https://codeskulptor-demos.commondatastorage.googleapis.com/descent/background%20music.mp3', 
            'https://pixabay.com/music/orchestral-cyberpunk-short-trailer-589196/', 
            'https://pixabay.com/music/alternative-hip-hop-battle-hip-hop-584304/',
            'https://pixabay.com/music/future-bass-that-future-bass-581055/'
]

t1=time.perf_counter()
def download_urls(audio_files):
    a_bytes=requests.get(a_url).content
    a_name=a_url.split('/')[4]
    a_name=f'{a_name}.mp3'
    with open (a_name,'wb') as a_file:
        a_file.write(a_bytes)
        print(f'{a_name} was downloaded..')

with concurrent.futures.ThreadPoolExecutor()as exe:
    exe.map(download_urls,audio_files)

t2=time.perf_counter()
print(f'Finished in {t2-t1} seconds')
```
