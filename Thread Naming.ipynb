Practice printing MainThread name and worker threads

```python
from threading import*
def display():
    print('Child Thread Name:', current_thread().getName())
    print('')
t=Thread(target=display)
t.start()
print('Main Thread Name:', current_thread().getName())
```

```python
from threading import *
def display(): # function is the task that runs in a new thread
    for i in range(10):
        print('Child Thread')
# Creation of Thread Object to Execute Display
t=Thread(target=display) # creates a thread and sets its job to run display()
t.start() # runs display() in parallel with the main thread
for i in range(10):
    print('Main Thread')
```

```python
import threading
import time

def worker():
    for i in range(5):
        print("Working thread:", threading.current_thread().name)
        time.sleep(1)

print("Main thread:", threading.current_thread().name)

t = threading.Thread(target=worker, name="WorkerThread")
t.start()

for i in range(5):
    print("Main thread working:", threading.current_thread().name)
    time.sleep(1)

t.join()
```
