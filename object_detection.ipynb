Without using class

```python


from ultralytics import YOLO
from google.colab import files
from IPython.display import Video, display
import glob

uploaded = files.upload()
video_name = list(uploaded.keys())[0]

model = YOLO("yolov8n.pt")

results = model.predict(
    source=video_name,
    save=True,
    conf=0.4
)

detected_objects = set()

for result in results:
    for box in result.boxes:
        class_id = int(box.cls[0])
        detected_objects.add(model.names[class_id])

print("\nObjects detected in the video:")
for obj in sorted(detected_objects):
    print("•", obj)

output_videos = glob.glob("/content/runs/detect/predict/*.mp4")

print("\nOutput Video with object detected ")
```

Using class object

```python
from ultralytics import YOLO
from google.colab import files
from IPython.display import Video, display
import glob


class ObjectDetector:

    def __init__(self):
        self.model = YOLO("yolov8n.pt")
        self.detected_objects = set()

    def upload_video(self):
        uploaded = files.upload()
        self.video_name = list(uploaded.keys())[0]
        print("Uploaded:", self.video_name)

    def detect_objects(self):
        results = self.model.predict(
            source=self.video_name,
            save=True,
            conf=0.4
        )

        for result in results:
            for box in result.boxes:
                class_id = int(box.cls[0])
                object_name = self.model.names[class_id]
                self.detected_objects.add(object_name)

        print("\nObjects Detected:")
        for obj in sorted(self.detected_objects):
            print("•", obj)

    def show_output(self):
        output_videos = glob.glob(
            "/content/runs/detect/predict/*.mp4"
        )

        print("\nDetected Output Video:")
        display(Video(output_videos[0], embed=True, width=900))


detector = ObjectDetector()

detector.upload_video()
detector.detect_objects()
detector.show_output()
```

Instead of writing all the code separately, we group related operations inside the ObjectDetector class.

upload_video() -- This function uploads the video from the user's computer.

detect_objects() -- This function performs the main object detection process.

show_output() -- This function finds the output video and displays it inside Google Colab.
